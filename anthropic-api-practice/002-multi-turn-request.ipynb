```python
#load environment variables

from dotenv import load_dotenv
load_dotenv()
```

```python
# create client and set model
from anthropic import Anthropic

client = Anthropic()
model = "claude-sonnet-4-6"
```

```python
# add function to add user message
def add_user_message(messages, content):
    messages.append({
        "role": "user",
        "content": content
    })
    return messages

# add assistant message to history
def add_assistant_message(messages, content):
    messages.append({
        "role": "assistant",
        "content": content
    })
    return messages

#  wrap this in function so we can call it pass messages as param and return  message content
def send_message(messages):
    return client.messages.create(
        model=model,
        max_tokens=1000,
        messages=messages
    )
    
# demo of multi turn conversation with history
messages = []
messages = add_user_message(messages, "What is quantum computing? Answer in one sentence")
# turn 1 user message
message = send_message(messages)
print(message.content[0].text)
messages = add_assistant_message(messages, message.content[0].text)

# turn 2 new user message
messages = add_user_message(messages, "Can you explain it in two sentences?")
message = send_message(messages)
print(message.content[0].text)
messages = add_assistant_message(messages, message.content[0].text)
print(messages)

```
