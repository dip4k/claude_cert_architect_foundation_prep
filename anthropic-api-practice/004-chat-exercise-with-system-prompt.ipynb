```python
#load environment variables

from dotenv import load_dotenv
load_dotenv()
```

```python
# create client and set model
from anthropic import Anthropic

client = Anthropic()
model = "claude-sonnet-4-6"
```

```python
# add function to add user message
def add_user_message(messages, content):
    messages.append({
        "role": "user",
        "content": content
    })
    return messages

# add assistant message to history
def add_assistant_message(messages, content):
    messages.append({
        "role": "assistant",
        "content": content
    })
    return messages

#  wrap this in function so we can call it pass messages and system as param and return  message content, set system prompt as well
def send_message(messages, system=None):
    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages,
    }
    
    if system:
        params["system"] = system
    
    message = client.messages.create(**params)
    return message
```

```python
# want to continue chat with system prompt included, want to give student hint from system prompt config, ai will not return answer immediately but a hint and we need to have chat function as earlier

messages = []

# With system prompt
system = """
You are a patient math tutor.
Do not directly answer a student's questions.
Guide them to a solution step by step.
"""

while True:
    user_input = input("You: ")
    print("You:", user_input)
    if user_input == "" or user_input.lower() in ["exit", "0"]:
        break
    messages = add_user_message(messages, user_input)
    message = send_message(messages, system)
    print("Assistant:", message.content[0].text)
    messages = add_assistant_message(messages, message.content[0].text)
    print("Messages so far:", messages)
```
