```python
#load environment variables

from dotenv import load_dotenv
load_dotenv()
```

```python
# create client and set model
from anthropic import Anthropic

client = Anthropic()
model = "claude-sonnet-4-6"
```

```python
# add function to add user message
def add_user_message(messages, content):
    messages.append({
        "role": "user",
        "content": content
    })
    return messages

# add assistant message to history
def add_assistant_message(messages, content):
    messages.append({
        "role": "assistant",
        "content": content
    })
    return messages

#  wrap this in function so we can call it pass messages as param and return  message content
def send_message(messages):
    return client.messages.create(
        model=model,
        max_tokens=1000,
        messages=messages
    )
    
# create a chat system with user input and assistant responses add exit when user enters exit or 0
messages = []
while True:
    user_input = input("You: ")
    print("You:", user_input)
    # break when user input is empty on press escape or enters "exit" or "0"
    if user_input == "" or user_input.lower() in ["exit", "0"]:
        break
    messages = add_user_message(messages, user_input)
    message = send_message(messages)
    print("Assistant:", message.content[0].text)
    messages = add_assistant_message(messages, message.content[0].text)


```
